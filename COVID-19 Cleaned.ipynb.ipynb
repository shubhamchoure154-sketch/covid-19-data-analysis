# COVID-19 Global Data Analysis Using Python

## Introduction

COVID-19 is a global pandemic that affected millions of people worldwide. 
Analyzing COVID-19 data helps understand the spread of the virus, recovery trends, 
and the severity of the pandemic across different countries.

In this project, we analyze a global COVID-19 dataset using Python. 
The goal is to explore the dataset, calculate important KPIs, and visualize trends 
such as confirmed cases, deaths, and recoveries across different countries and over time.

## Project Objectives

The main objectives of this project are:

- To perform exploratory data analysis (EDA) on the COVID-19 dataset.
- To identify trends in confirmed cases, deaths, and recoveries.
- To calculate key performance indicators (KPIs) such as death rate and recovery rate.
- To visualize the spread of COVID-19 using different charts and graphs.
- To compare the impact of COVID-19 across different countries.

## Dataset Description

The dataset used in this project contains global COVID-19 statistics. 
It includes daily records of confirmed cases, deaths, and recovered cases 
from different countries and regions.

Main columns in the dataset include:

- **Date** – The date of the recorded data.
- **Province/State** – The state or province (if available).
- **Country/Region** – The country where the cases were reported.
- **Latitude and Longitude** – Geographic location of the region.
- **Confirmed** – Total number of confirmed COVID-19 cases.
- **Deaths** – Total number of deaths caused by COVID-19.
- **Recovered** – Total number of recovered cases.

## Technologies Used

The following tools and libraries were used in this project:

- **Python** – Programming language used for data analysis.
- **Pandas** – Used for data cleaning and data manipulation.
- **NumPy** – Used for numerical computations.
- **Matplotlib** – Used for creating visualizations.
- **Seaborn** – Used for advanced data visualization.
- **Jupyter Notebook** – Used for writing and running the analysis.

## Key Performance Indicators (KPIs)

In this project, several KPIs are calculated to better understand the pandemic:

- **Total Confirmed Cases** – Total number of reported COVID-19 cases.
- **Total Deaths** – Total number of deaths due to COVID-19.
- **Total Recovered Cases** – Total number of people who recovered.
- **Active Cases** – Current active cases calculated as:

Active Cases = Confirmed − Deaths − Recovered

- **Death Rate** – Percentage of deaths among confirmed cases.
- **Recovery Rate** – Percentage of recovered cases among confirmed cases.

## Visualization

## Top 10 Countries with Highest Cases

```python
latest = df[df['Date'] == df['Date'].max()]

top10 = latest.groupby('Country/Region')['Confirmed'].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,6))

top10.plot(kind='bar')

plt.title("Top 10 Countries by Confirmed Cases")
plt.xlabel("Country")
plt.ylabel("Confirmed Cases")

plt.show()
```

```python
latest = df[df['Date'] == df['Date'].max()]

top10 = latest.groupby('Country/Region')['Confirmed'].sum().sort_values(ascending=True).tail(10)

plt.figure(figsize=(10,6))
top10.plot(kind='barh')

plt.title("Top 10 Countries by Confirmed Cases")
plt.xlabel("Confirmed Cases")
plt.ylabel("Country")

plt.show()
```

### Insight:
The chart shows that a small number of countries have the highest confirmed COVID-19 cases.  
This indicates that the spread of the virus was more intense in specific regions rather than evenly distributed.  
Factors like population size, urbanization, and international travel contributed to higher case numbers in these countries.

## Death vs Recovery Comparison

```python
latest = df[df['Date'] == df['Date'].max()]

data = latest.groupby('Country/Region')[['Deaths','Recovered']].sum().sort_values(by='Deaths', ascending=False).head(10)

data.plot(kind='bar', figsize=(10,6))

plt.title("Deaths vs Recovery (Top Countries)")
plt.show()
```

### Insight:
The graph shows that recovery cases are significantly higher than death cases across the top affected countries.  
This indicates that despite the high number of infections, a large proportion of patients were able to recover.  
It also reflects the effectiveness of healthcare systems and medical responses during the pandemic.

## Active Cases Trend

```python
active_cases = df.groupby('Date')['Active'].sum()

plt.figure(figsize=(12,6))

active_cases.plot()

plt.title("Active Cases Over Time")
plt.xlabel("Date")
plt.ylabel("Active Cases")

plt.show()
```

### Insight:
The graph shows that active COVID-19 cases increased significantly over time, indicating rapid spread of the virus during peak periods.  
Fluctuations in the trend suggest multiple waves of infection.  
A decline in active cases in later stages reflects recovery improvements and control measures taken by governments.

## Correlation Heatmap

```python
corr = df[['Confirmed','Deaths','Recovered','Active']].corr()

plt.figure(figsize=(8,6))

sns.heatmap(corr, annot=True)

plt.title("Correlation Between Variables")

plt.show()
```

### Insight:
The heatmap shows a strong positive correlation between confirmed, deaths, recovered, and active cases.  
This means that as the number of confirmed cases increases, deaths and recoveries also increase proportionally.  
The strong relationships indicate consistent patterns in the spread and impact of COVID-19.

## Top 10 Countries by Death Rate

```python
latest['Death Rate'] = latest['Deaths'] / latest['Confirmed'] * 100

death_rate = latest.groupby('Country/Region')['Death Rate'].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,6))

death_rate.plot(kind='bar')

plt.title("Top 10 Countries by Death Rate")
plt.ylabel("Death Rate %")

plt.show()
```

### Insight:
The chart highlights that some countries have a higher death rate despite not always having the highest number of confirmed cases.  
This indicates differences in healthcare quality, population demographics, and response strategies.  
Higher death rates may reflect challenges in medical infrastructure or delayed interventions during the pandemic.

## KPI Section for COVID-19 Project

## Calculate Main KPIs

```python
latest = df[df['Date'] == df['Date'].max()]

total_confirmed = latest['Confirmed'].sum()
total_deaths = latest['Deaths'].sum()
total_recovered = latest['Recovered'].sum()
total_active = latest['Active'].sum()

death_rate = (total_deaths / total_confirmed) * 100
recovery_rate = (total_recovered / total_confirmed) * 100

print("Total Confirmed Cases:", total_confirmed)
print("Total Deaths:", total_deaths)
print("Total Recovered:", total_recovered)
print("Total Active Cases:", total_active)
print("Death Rate:", round(death_rate,2), "%")
print("Recovery Rate:", round(recovery_rate,2), "%")
```

## KPI Bar Chart

```python
kpi_data = [total_confirmed, total_deaths, total_recovered, total_active]

labels = ['Confirmed', 'Deaths', 'Recovered', 'Active']

plt.figure(figsize=(8,5))
plt.bar(labels, kpi_data)

plt.title("COVID-19 Global KPIs")
plt.ylabel("Number of Cases")

plt.show()
```

## KPI Pie Chart

```python
pie_data = [total_deaths, total_recovered, total_active]

labels = ['Deaths', 'Recovered', 'Active']

plt.figure(figsize=(6,6))
plt.pie(pie_data, labels=labels, autopct='%1.1f%%')

plt.title("COVID Case Distribution")

plt.show()
```

### Insight:
The KPI chart shows that total confirmed cases are significantly higher than deaths, while recovered cases form a major portion of total cases.  
This indicates that although the pandemic spread widely, recovery rates were relatively strong.  
Active cases highlight the ongoing impact and burden on healthcare systems during the pandemic.

## KPI Trend (Daily New Cases)

```python
global_cases = df.groupby('Date')['Confirmed'].sum()

daily_cases = global_cases.diff()

plt.figure(figsize=(12,6))
plt.plot(daily_cases)

plt.title("Daily New COVID Cases")
plt.xlabel("Date")
plt.ylabel("New Cases")

plt.show()
```

### Insight:
The graph shows fluctuations in daily new COVID-19 cases, highlighting periods of rapid increase and temporary declines.  
Peaks in daily cases represent waves of infection, while drops indicate the effect of control measures or recovery trends.  
Analyzing daily cases helps understand the short-term dynamics of the pandemic beyond cumulative totals.

## Boxplot (Distribution of Cases)

```python
plt.figure(figsize=(8,6))

sns.boxplot(data=df[['Confirmed','Deaths','Recovered']])

plt.title("Distribution of COVID Variables")

plt.show()
```

### Insight:
The boxplot shows that most regions have relatively low numbers of confirmed, death, and recovered cases, while a few regions have extremely high values.  
This indicates the presence of outliers and a highly skewed distribution in the dataset.  
Such variation highlights that COVID-19 impact was uneven across different countries and regions.

## Continuous vs Continuous (Scatter Plot)

```python
plt.figure(figsize=(8,6))

sns.scatterplot(x='Confirmed', y='Deaths', color='green', data=df )

plt.title("Confirmed vs Deaths")

plt.show() 
```

### Insight:
The scatter plot shows a strong positive relationship between confirmed and death cases.  
As confirmed cases increase, deaths also tend to increase proportionally.  
This highlights the direct impact of infection spread on mortality and emphasizes the severity in highly affected regions.

## Sunburst Chart

```python
latest = df[df['Date'] == df['Date'].max()]

fig = px.sunburst(latest,
                  path=['Country/Region'],
                  values='Confirmed',
                  title='COVID Cases by Country')

fig.show()
```

### Insight:
The sunburst chart shows the distribution of confirmed COVID-19 cases across countries, highlighting which countries contribute most to the global case count.  
It reveals that a few countries account for a large proportion of cases, while many others have relatively lower numbers.  
This visualization helps understand the hierarchical spread of COVID-19 worldwide.

## Violin Plot

```python
plt.figure(figsize=(8,6))

sns.violinplot(data=df[['Confirmed','Deaths','Recovered']])

plt.title("Violin Plot of COVID Variables")

plt.show()
```

### Insight:
The violin plot shows that most regions have relatively low numbers of confirmed, death, and recovered cases, while a few regions have extremely high values.  
This indicates a skewed distribution and the presence of outliers, reflecting uneven impact of COVID-19 across countries.  
The spread of the distribution also helps identify variability in case numbers among affected regions.

## Swarm Plot

```python
plt.figure(figsize=(8,6))

sns.swarmplot(x='Country/Region', y='Deaths', data=latest.head(15))

plt.xticks(rotation=90)

plt.title("Deaths Distribution by Country")

plt.show()
```

### Insight:
The plot highlights the top 15 countries with the highest COVID-19 death counts. 
It clearly shows that a few countries are heavily affected, with significantly higher deaths 
compared to others, indicating a concentrated impact of the pandemic.
